```python
from pathlib import Path

out_root = Path("outputs_demo_notebook") / "DEMO"
run_dir = max(out_root.iterdir(), key=lambda p: p.stat().st_mtime)
run_dir
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

img = mpimg.imread(run_dir / "plots" / "spectrum.png")
plt.figure(figsize=(10, 4))
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
import json

results = json.loads((run_dir / "reports" / "results.json").read_text(encoding="utf-8"))
list(results.keys())
```

```python
molecules = results.get("detections", {})
summary = [
    {
        "molecule": k,
        "status": v.get("status"),
        "confidence": float(v.get("confidence", 0.0)),
        "max_snr": float(v.get("max_snr", 0.0)),
        "fap": float(v.get("fap", 1.0)),
    }
    for k, v in molecules.items()
]
sorted(summary, key=lambda r: (r["confidence"], r["max_snr"]), reverse=True)[:10]
```
